## 3. Desarrollo

### 3.1 Definiciones recopiladas

Se recopilaron seis definiciones de fuentes diversas: dos académicas clásicas y cuatro generadas por modelos de lenguaje de gran escala (LLMs).

---

**Fuente 1 — John McCarthy (1956, Conferencia de Dartmouth):**
> "La ciencia e ingeniería de hacer máquinas inteligentes, especialmente programas informáticos inteligentes."

McCarthy acuñó el término *Artificial Intelligence* en 1955 al proponer el proyecto de investigación de verano de Dartmouth. Su definición es fundacional y orientada a la ingeniería.

---

**Fuente 2 — Russell & Norvig (2020, *Artificial Intelligence: A Modern Approach*, 4ª ed.):**
> "El estudio de agentes que reciben percepciones del entorno y ejecutan acciones. Cada agente implementa una función que mapea secuencias de percepciones a acciones."

Russell y Norvig organizan las definiciones de IA en cuatro categorías: sistemas que piensan como humanos, que piensan racionalmente, que actúan como humanos y que actúan racionalmente. Su propia postura se alinea con la racionalidad del agente.

---

**Fuente 3 — ChatGPT (OpenAI, GPT-4o, marzo 2026):**
> "La Inteligencia Artificial es la rama de la informática que desarrolla sistemas capaces de realizar tareas que normalmente requieren inteligencia humana, como el razonamiento, el aprendizaje, la percepción y la toma de decisiones."

---

**Fuente 4 — Gemini (Google, marzo 2026):**
> "La IA es el campo de la ciencia computacional dedicado a crear sistemas que simulan procesos cognitivos humanos, incluyendo el aprendizaje automático, el procesamiento del lenguaje natural y la resolución de problemas complejos."

---

**Fuente 5 — Claude (Anthropic, marzo 2026):**
> "La Inteligencia Artificial es el estudio y desarrollo de sistemas computacionales capaces de realizar tareas que, al ser ejecutadas por humanos, se considerarían inteligentes; abarca desde el razonamiento lógico hasta el aprendizaje a partir de datos."

---

**Fuente 6 — Copilot (Microsoft, marzo 2026):**
> "La IA es la capacidad de las máquinas para imitar funciones cognitivas humanas como aprender, razonar y resolver problemas, habilitada mediante algoritmos y grandes volúmenes de datos."

### 3.3 Análisis: similitudes y diferencias

**Similitudes:**

- Todas las definiciones coinciden en que la IA involucra la realización de tareas asociadas a la inteligencia, ya sea humana o racional.
- La mayoría hace referencia explícita o implícita al razonamiento y al aprendizaje como capacidades centrales.
- Ninguna definición niega que la IA sea un campo de la computación o la ingeniería; hay consenso en su naturaleza tecnológica.

**Diferencias:**

- McCarthy define la IA desde la *construcción* (qué se hace), mientras que Russell & Norvig la definen desde el *comportamiento* del sistema resultante (cómo actúa).
- Los LLMs tienden a definir la IA por enumeración de capacidades (aprendizaje, razonamiento, percepción), sin establecer un principio unificador. Esto hace sus definiciones más amplias pero menos precisas.
- Russell & Norvig son los únicos que formalizan la definición mediante el concepto de *agente* y *función de mapeo*, lo que otorga mayor rigor operacional.
- Claude es el único LLM que introduce el criterio de *juicio humano* ("tareas que se considerarían inteligentes"), lo que acerca su definición al Test de Turing de Turing (1950).
- Copilot es el único que menciona explícitamente los *datos* como componente habilitador, reflejando la influencia del paradigma actual de aprendizaje profundo.

## 5. Resultado

### 5.1 Definición propia

> **"La Inteligencia Artificial es la disciplina que estudia y construye sistemas capaces de percibir su entorno, razonar sobre la información disponible y actuar de manera que maximicen el logro de objetivos definidos; ya sea emulando capacidades cognitivas humanas o superándolas mediante métodos formales, estadísticos o basados en datos."**

### 5.2 Justificación

Esta definición integra deliberadamente tres dimensiones presentes en las fuentes analizadas:

- **Dimensión constructiva** (McCarthy): la IA no solo se estudia, se *construye*.
- **Dimensión comportamental** (Russell & Norvig): el sistema *percibe*, *razona* y *actúa* con un objetivo, lo que lo convierte en un agente.
- **Dimensión metodológica** (Copilot, LLMs): se reconoce que los métodos pueden ser formales, estadísticos o basados en datos, evitando restringir la definición a un paradigma técnico específico.

Se evitó la referencia exclusiva a la inteligencia *humana* como criterio de comparación, ya que esto limita el alcance de la IA a la imitación y excluye sistemas que superan capacidades humanas en dominios específicos (como el ajedrez o el diagnóstico médico por imagen).

## 6. Conclusión

El ejercicio comparativo evidencia que la IA carece de una definición unívoca porque su alcance ha evolucionado junto con sus métodos y aplicaciones. Las definiciones académicas tienden a ser más precisas y operacionalizables, mientras que las generadas por LLMs son más amplias y accesibles pero menos rigurosas. La tensión central entre *construir* sistemas inteligentes (McCarthy) y *estudiar* el comportamiento racional (Russell & Norvig) refleja dos tradiciones que coexisten en el campo. Construir una definición propia obliga a tomar postura frente a esa tensión y a reconocer que la IA es, simultáneamente, una disciplina científica, una práctica de ingeniería y un campo en permanente redefinición.
